```python
# Inisialisasi Spark & Pembangkitan Data Log Jaringan Unik
!apt-get update -qq
!apt-get install -y openjdk-17-jdk-headless -qq
!pip install -q pyspark pandas matplotlib

import os
import pandas as pd
import numpy as np
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

# Membuat Spark Session dengan identitas unik ITERA
spark = SparkSession.builder \
    .appName("ITERA_CyberSecurity_Analysis") \
    .master("local[*]") \
    .getOrCreate()

# Pembangkitan dataset log keamanan jaringan (menggunakan seed berbeda agar orisinal)
np.random.seed(101)
n_records = 2500
timestamps = pd.date_range(start="2026-01-01", periods=n_records, freq="30s")

raw_data = {
    "PacketID": [f"PKT_{i:04d}" for i in range(n_records)],
    "EventTime": timestamps,
    "PayloadSize": np.random.randint(64, 1500, size=n_records),
    "ResponseTimeMs": np.random.uniform(5.0, 300.0, size=n_records),
    "ConnectionStatus": np.random.choice(["ALLOW", "BLOCK", "SUSPICIOUS"], size=n_records, p=[0.7, 0.2, 0.1]),
    "ThreatLevel": np.random.choice([0, 1], size=n_records, p=[0.8, 0.2])
}

df_source = pd.DataFrame(raw_data)
os.makedirs("logs_data", exist_ok=True)
df_source.to_csv("logs_data/network_traffic.csv", index=False)

# Membaca ke PySpark DataFrame
spark_df = spark.read.csv("logs_data/network_traffic.csv", header=True, inferSchema=True)
spark_df.printSchema()
spark_df.show(5)
```

```python
# Analisis Eksploratif Menggunakan Spark SQL Aggregation
spark_df.createOrReplaceTempView("network_logs")

summary_sql = spark.sql("""
    SELECT ConnectionStatus,
           COUNT(*) as TotalPackets,
           ROUND(AVG(PayloadSize), 2) as AvgPayload,
           ROUND(AVG(ResponseTimeMs), 2) as AvgResponseTime
    FROM network_logs
    GROUP BY ConnectionStatus
    ORDER BY TotalPackets DESC
""")
summary_sql.show()
```

```python
# Feature Engineering & Klasifikasi Menggunakan Decision Tree Classifier
from pyspark.ml.feature import VectorAssembler, StringIndexer
from pyspark.ml.classification import DecisionTreeClassifier
from pyspark.ml.evaluation import MulticlassClassificationEvaluator
import time

# Ekstraksi fitur waktu
processed_df = spark_df.withColumn("EventTime", F.to_timestamp("EventTime")) \
                       .withColumn("LogHour", F.hour("EventTime"))

# Indexing kolom kategorikal
status_indexer = StringIndexer(inputCol="ConnectionStatus", outputCol="Status_Index")
indexed_df = status_indexer.fit(processed_df).transform(processed_df)

# Merakit vektor fitur
feature_features = ["PayloadSize", "ResponseTimeMs", "LogHour", "Status_Index"]
assembler = VectorAssembler(inputCols=feature_features, outputCol="features")
vector_df = assembler.transform(indexed_df)

# Split Data Latih & Uji (Rasio 75:25)
train_data, test_data = vector_df.randomSplit([0.75, 0.25], seed=101)

# Pelatihan Model Decision Tree Classifier
start_time = time.time()
dt = DecisionTreeClassifier(labelCol="ThreatLevel", featuresCol="features", maxDepth=4)
dt_model = dt.fit(train_data)
predictions = dt_model.transform(test_data)
training_duration = time.time() - start_time

# Evaluasi Akurasi
evaluator = MulticlassClassificationEvaluator(labelCol="ThreatLevel", predictionCol="prediction", metricName="accuracy")
accuracy_score = evaluator.evaluate(predictions)

print(f"=== HASIL EKSPERIMEN DECISION TREE CYBERSECURITY ===")
print(f"Model Accuracy  : {accuracy_score:.4f}")
print(f"Training Time   : {training_duration:.4f} seconds")

# Menyimpan hasil ke CSV unik
metrics_pd = pd.DataFrame({
    "Algorithm": ["Decision Tree Classifier"],
    "Accuracy": [accuracy_score],
    "ExecutionTime_Sec": [training_duration]
})
metrics_pd.to_csv("hasil_cybersecurity_dt.csv", index=False)
```
